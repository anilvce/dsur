```r
# Install the package if you haven't already
install.packages("remotes")
remotes::install_github("mkearney/rmd2jupyter")

sum=0
a=c(1,3,5,10,7)
for( num in a){
  sum=sum+num
}
print(sum)
```

Maximum of numbers



```r
max=a[1]
print(a)
for (num in a){
 # print(num)
  if ( max <= num ) {
    max=num
  }
}
print(max)
print(max(a))
```

```r
# List of strike rates
b <- list(100, 123, 34, 99)
good<-c()
bad<-c()
if(b[1]>100){
  good<-c(good,1)
} else{
  bad<-c(bad,1)
}


if(b[2]>100){
  good<-c(good,2)
} else{
  bad<-c(bad,2)
}

i<-4
if(b[i]>100){
  good<-c(good,i)
} else{
  bad<-c(bad,i)
}

# Convert list to numeric vector for easier processing
strike_rates <- unlist(b)

# Categorize using ifelse
categories <- ifelse(strike_rates > 100, "good", "bad")

# Display results
data.frame(StrikeRate = strike_rates, Category = categories)
```
